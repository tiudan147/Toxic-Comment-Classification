# Thực nghiệm chứng minh lựa chọn tham số trong `train.py`

Notebook này chạy `test_tham_so.py` trên nhánh `quang/test_tham_so` để chứng minh cơ sở thực nghiệm cho 4 tham số đang dùng trong `train.py`:
- `max_features = 5000` (TF-IDF Vectorizer)
- `alpha = 0.5` (Multinomial Naive Bayes)
- `C = 1.0` (Linear SVM với `class_weight='balanced'`)
- `vector_size = 150` (Word2Vec + Linear SVM)

**Hướng dẫn dữ liệu:** Nén thư mục `dataset/processed_data` (gồm `train.csv` và `valid.csv`) thành `processed_data.zip` rồi upload lên tab **Files** bên trái Colab trước khi chạy Cell 3.

```python
import os, zipfile
os.makedirs("dataset/processed_data", exist_ok=True)

zip_candidates = ["/content/processed_data.zip", "processed_data.zip"]
found = next((p for p in zip_candidates if os.path.exists(p)), None)
if found:
    with zipfile.ZipFile(found, "r") as zf:
        zf.extractall("dataset/processed_data")
    # Nếu giải nén bị lồng thêm thư mục processed_data/ thì dọn ra ngoài
    nested = "dataset/processed_data/processed_data"
    if os.path.isdir(nested):
        import shutil
        for f in os.listdir(nested):
            shutil.move(os.path.join(nested, f), os.path.join("dataset/processed_data", f))
    print("Đã giải nén dataset/processed_data:", os.listdir("dataset/processed_data"))
else:
    print("Chưa thấy processed_data.zip. Hãy upload file hoặc đặt train.csv, valid.csv vào dataset/processed_data/")
```
